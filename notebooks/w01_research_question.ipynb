## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# Total impressions across all content
print(f"Total impressions in starter data: {df['impressions_90d'].sum():,}")

# Average clicks per page for content with >0 impressions
df_active = df[df['impressions_90d'] > 0]
print(f"Average clicks per page: {df_active['clicks_90d'].mean():.2f}")

# Correlation between sessions and impressions
corr = df_active['sessions_90d'].corr(df_active['impressions_90d'])
print(f"Correlation between sessions and impressions: {corr:.2f}")
```
